```python
import pandas as pd
df = pd.read_csv("medical_dataset_1100_rows_v2.csv")
print(df)
print(df.head())
print(df.info())
print(df.describe())
```

```python
import pandas as pd


print("===== SHAPE =====")
print(df.shape)


print("\n===== DATA TYPES =====")
print(df.dtypes)


print("\n===== FILTER COLUMNS =====")
print(df[["Age", "Gender", "Disease"]])


print("\n===== AGE > 50 =====")
print(df[df["Age"] > 50])

print("\n===== DISEASE = Diabetes =====")
print(df[df["Disease"] == "Diabetes"])

print("\n===== GENDER = Female =====")
print(df[df["Gender"] == "Female"])

# Multiple Conditions
print("\n===== AGE > 50 AND GENDER = Male =====")
print(df[(df["Age"] > 50) & (df["Gender"] == "Male")])


print("\n===== SORT BY AGE =====")
print(df.sort_values(by="Age"))


print("\n===== SORT BY AGE DESC =====")
print(df.sort_values(by="Age", ascending=False))


print("\n===== SORT BY GENDER AND AGE =====")
print(df.sort_values(by=["Gender", "Age"]))
```

```python
print(df["Age"].corr(df["Weight_kg"]))

print(df["BMI"].corr(df["Blood_Sugar"]))
```

```python
print(df.corr(numeric_only=True, method='pearson'))

print(df.corr(numeric_only=True, method='spearman'))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# =====================================
# View Data
# =====================================

print("First 5 Rows")
print(df.head())

print("\nShape")
print(df.shape)

print("\nData Types")
print(df.dtypes)

# =====================================
# Filter Data
# =====================================

print("\nPatients with Age > 50")
age_filter = df[df["Age"] > 50]
print(age_filter)

print("\nPatients with Diabetes")
diabetes_patients = df[df["Disease"] == "Diabetes"]
print(diabetes_patients)

print("\nFemale Patients")
female_patients = df[df["Gender"] == "Female"]
print(female_patients)

# =====================================
# Sort Data
# =====================================

print("\nSorted by Age")
sorted_df = df.sort_values(by="Age", ascending=False)
print(sorted_df)

# =====================================
# Statistics
# =====================================

print("\nMean Age =", df["Age"].mean())
print("Median Age =", df["Age"].median())

print("\nMode Age")
print(df["Age"].mode())

print("Standard Deviation Age =", df["Age"].std())

# =====================================
# Summary Statistics
# =====================================

print("\nSummary Statistics")
print(df.describe())

# =====================================
# Group By
# =====================================

print("\nAverage Age by Disease")
disease_avg = df.groupby("Disease")["Age"].mean()
print(disease_avg)

# =====================================
# Correlation
# =====================================

print("\nCorrelation")
corr = df[["Age", "Blood_Sugar"]].corr()
print(corr)

# =====================================
# Scatter Plot
# =====================================

plt.figure(figsize=(6,4))
plt.scatter(df["Age"], df["Blood_Sugar"])
plt.title("Age vs Blood Sugar")
plt.xlabel("Age")
plt.ylabel("Blood Sugar")
plt.grid(True)
plt.show()

# =====================================
# Bar Plot
# =====================================

disease_avg.plot(kind="bar")
plt.title("Average Age by Disease")
plt.xlabel("Disease")
plt.ylabel("Average Age")
plt.show()
```
