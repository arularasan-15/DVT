```python
import pandas as pd

# ==================================================
# HEALTHCARE DATASET CLEANING
# ==================================================

# 1. Load dataset
df = pd.read_csv("healthcare_dataset.csv")

# Make a copy of original dataset
healthcare = df.copy()


# --------------------------------------------------
# 2. Check original data
# --------------------------------------------------

print("========================================")
print("ORIGINAL HEALTHCARE DATA")
print("========================================")

print("Original rows:", len(healthcare))
print("Original columns:", len(healthcare.columns))

print("\nColumn names:")
print(healthcare.columns.tolist())

print("\nMissing values:")
print(healthcare.isnull().sum())


# --------------------------------------------------
# 3. Standardize column names
# --------------------------------------------------

healthcare.columns = (
    healthcare.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_", regex=False)
)

print("\nStandardized column names:")
print(healthcare.columns.tolist())


# --------------------------------------------------
# 4. Remove duplicate records
# --------------------------------------------------

duplicate_count = healthcare.duplicated().sum()

print("\nDuplicate records:", duplicate_count)

healthcare.drop_duplicates(inplace=True)


# --------------------------------------------------
# 5. Handle missing values
# --------------------------------------------------

# Numerical columns
numeric_columns = [
    "age",
    "billing_amount",
    "room_number"
]

for col in numeric_columns:
    if col in healthcare.columns:
        healthcare[col] = healthcare[col].fillna(
            healthcare[col].median()
        )


# Categorical columns
categorical_columns = [
    "name",
    "gender",
    "blood_type",
    "medical_condition",
    "doctor",
    "hospital",
    "insurance_provider",
    "admission_type",
    "medication",
    "test_results"
]

for col in categorical_columns:
    if col in healthcare.columns:
        healthcare[col] = healthcare[col].fillna("Unknown")


# Date columns
date_columns = [
    "date_of_admission",
    "discharge_date"
]

for col in date_columns:
    if col in healthcare.columns:
        healthcare[col] = pd.to_datetime(
            healthcare[col],
            errors="coerce"
        )


# --------------------------------------------------
# 6. Create useful derived columns
# --------------------------------------------------

# Patient stay duration
healthcare["length_of_stay"] = (
    healthcare["discharge_date"]
    - healthcare["date_of_admission"]
).dt.days


# Age category
healthcare["age_category"] = pd.cut(
    healthcare["age"],
    bins=[0, 18, 35, 50, 65, 100],
    labels=[
        "Child",
        "Young Adult",
        "Adult",
        "Senior Adult",
        "Elderly"
    ]
)


# Billing category
healthcare["billing_category"] = pd.cut(
    healthcare["billing_amount"],
    bins=[-float("inf"), 10000, 25000, 40000, float("inf")],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)


# Admission category
healthcare["admission_category"] = healthcare["admission_type"].astype(str).str.title()


# --------------------------------------------------
# 7. Clean text columns
# --------------------------------------------------

text_columns = [
    "name",
    "gender",
    "blood_type",
    "medical_condition",
    "doctor",
    "hospital",
    "insurance_provider",
    "admission_type",
    "medication",
    "test_results"
]

for col in text_columns:
    if col in healthcare.columns:
        healthcare[col] = (
            healthcare[col]
            .astype(str)
            .str.strip()
        )


# --------------------------------------------------
# 8. Check cleaned data
# --------------------------------------------------

print("\n========================================")
print("CLEANED HEALTHCARE DATA")
print("========================================")

print("Original rows:", len(df))
print("Cleaned rows:", len(healthcare))
print("Columns:", len(healthcare.columns))

print("\nMissing values after cleaning:")

missing = healthcare.isnull().sum()

if missing.sum() == 0:
    print("No missing values")
else:
    print(missing[missing > 0])


# --------------------------------------------------
# 9. Display cleaned dataset
# --------------------------------------------------

print("\nFirst 5 rows:")
display(healthcare.head())


# --------------------------------------------------
# 10. Dataset information
# --------------------------------------------------

print("\n========================================")
print("DATASET INFORMATION")
print("========================================")

healthcare.info()


# --------------------------------------------------
# 11. Statistical summary
# --------------------------------------------------

print("\n========================================")
print("STATISTICAL SUMMARY")
print("========================================")

display(healthcare.describe())


# --------------------------------------------------
# 12. Final shape
# --------------------------------------------------

print("\nFinal dataset shape:", healthcare.shape)
```

```python
import sqlite3
import pandas as pd

# ==================================================
# HEALTHCARE SQL DATABASE
# ==================================================

# 1. Create an in-memory SQLite database
conn = sqlite3.connect(":memory:")
cursor = conn.cursor()


# --------------------------------------------------
# 2. Read and execute SQL file
# --------------------------------------------------

with open("healthcare_dataset.sql", "r") as file:
    sql_script = file.read()

cursor.executescript(sql_script)


# --------------------------------------------------
# 3. Display tables
# --------------------------------------------------

tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

print("===== TABLES IN DATABASE =====")
print(tables)


# --------------------------------------------------
# 4. Get first table name
# --------------------------------------------------

if len(tables) > 0:

    table_name = tables.iloc[0, 0]

    print("\nSelected Table:", table_name)


    # --------------------------------------------------
    # 5. Read table into DataFrame
    # --------------------------------------------------

    df = pd.read_sql_query(
        f'SELECT * FROM "{table_name}"',
        conn
    )


    # --------------------------------------------------
    # 6. Display first 5 records
    # --------------------------------------------------

    print("\n===== HEAD =====")
    print(df.head())


    # --------------------------------------------------
    # 7. Display last 5 records
    # --------------------------------------------------

    print("\n===== TAIL =====")
    print(df.tail())


    # --------------------------------------------------
    # 8. Display shape
    # --------------------------------------------------

    print("\n===== SHAPE =====")
    print(df.shape)


    # --------------------------------------------------
    # 9. Display column names
    # --------------------------------------------------

    print("\n===== COLUMNS =====")
    print(df.columns.tolist())


    # --------------------------------------------------
    # 10. Display dataset information
    # --------------------------------------------------

    print("\n===== INFO =====")
    df.info()


    # --------------------------------------------------
    # 11. Display statistical summary
    # --------------------------------------------------

    print("\n===== DESCRIBE =====")
    print(df.describe(include="all"))


    # --------------------------------------------------
    # 12. Missing values
    # --------------------------------------------------

    print("\n===== MISSING VALUES =====")
    print(df.isnull().sum())


    # --------------------------------------------------
    # 13. Duplicate records
    # --------------------------------------------------

    print("\n===== DUPLICATES =====")
    print("Duplicate rows:", df.duplicated().sum())

else:
    print("No tables found in the SQL database.")


# --------------------------------------------------
# 14. Close database
# --------------------------------------------------

conn.close()
```
