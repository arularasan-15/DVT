```python
import pandas as pd
df = pd.read_csv("platform_statistics_2026.csv")

# ==================================================
# PLATFORM STATISTICS DATA CLEANING
# ==================================================

# 1. Make a copy of the original dataset
platform = df.copy()


# --------------------------------------------------
# 2. Check original data
# --------------------------------------------------

print("Original rows:", len(platform))
print("Original columns:", len(platform.columns))

print("\nMissing values:")
print(platform.isnull().sum())


# --------------------------------------------------
# 3. Handle missing values
# --------------------------------------------------

# Numerical columns
numeric_columns = [
    "monthly_active_users_billions",
    "year_over_year_growth_pct",
    "avg_daily_time_minutes",
    "avg_engagement_rate_pct",
    "social_commerce_adoption_pct",
    "founded_year"
]

for col in numeric_columns:
    if col in platform.columns:
        platform[col] = platform[col].fillna(platform[col].median())


# Categorical columns
categorical_columns = [
    "platform",
    "primary_age_group",
    "primary_content_format",
    "headquarters_country",
    "data_source"
]

for col in categorical_columns:
    if col in platform.columns:
        platform[col] = platform[col].fillna("Unknown")


# --------------------------------------------------
# 4. Remove duplicate records
# --------------------------------------------------

platform.drop_duplicates(inplace=True)


# --------------------------------------------------
# 5. Create useful derived columns
# --------------------------------------------------

# Estimated daily active user value
platform["estimated_daily_users_billions"] = (
    platform["monthly_active_users_billions"] * 0.30
)


# Engagement score
platform["engagement_score"] = (
    platform["avg_engagement_rate_pct"] *
    platform["avg_daily_time_minutes"]
)


# Growth category
platform["growth_category"] = pd.cut(
    platform["year_over_year_growth_pct"],
    bins=[-float("inf"), 5, 10, float("inf")],
    labels=["Low Growth", "Medium Growth", "High Growth"]
)


# Social commerce category
platform["commerce_category"] = pd.cut(
    platform["social_commerce_adoption_pct"],
    bins=[-float("inf"), 20, 40, float("inf")],
    labels=["Low", "Medium", "High"]
)


# --------------------------------------------------
# 6. Calculate platform age
# --------------------------------------------------

current_year = 2026

platform["platform_age"] = (
    current_year - platform["founded_year"]
)


# --------------------------------------------------
# 7. Standardize column names
# --------------------------------------------------

platform.columns = (
    platform.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_", regex=False)
)


# --------------------------------------------------
# 8. Check cleaned data
# --------------------------------------------------

print("\n========================================")
print("CLEANED PLATFORM DATA")
print("========================================")

print("Original rows:", len(df))
print("Cleaned rows:", len(platform))
print("Columns:", len(platform.columns))

print("\nMissing values after cleaning:")
missing = platform.isnull().sum()

if missing.sum() == 0:
    print("No missing values")
else:
    print(missing[missing > 0])


# --------------------------------------------------
# 9. Display cleaned dataset
# --------------------------------------------------

print("\nFirst 5 rows:")
display(platform.head())


# --------------------------------------------------
# 10. Dataset information
# --------------------------------------------------

print("\nDataset Information:")
platform.info()
```

```python
import sqlite3
import pandas as pd

# --------------------------------------------------
# 1. Make platform DataFrame from df
# --------------------------------------------------

platform = df.copy()

# --------------------------------------------------
# 2. Create / connect SQLite database
# --------------------------------------------------

conn = sqlite3.connect("platform_analytics.db")

# --------------------------------------------------
# 3. Export DataFrame to SQLite
# --------------------------------------------------

platform.to_sql(
    "platform_statistics",
    conn,
    if_exists="replace",
    index=False
)

print("Platform data exported successfully!")

# --------------------------------------------------
# 4. Check number of rows
# --------------------------------------------------

result = pd.read_sql(
    "SELECT COUNT(*) AS total_rows FROM platform_statistics",
    conn
)

print("\nTotal rows:")
display(result)

# --------------------------------------------------
# 5. Display first 5 rows
# --------------------------------------------------

result = pd.read_sql(
    "SELECT * FROM platform_statistics LIMIT 5",
    conn
)

print("\nFirst 5 rows:")
display(result)

# --------------------------------------------------
# 6. Close connection
# --------------------------------------------------

conn.close()

print("\nSQLite database connection closed.")
```
