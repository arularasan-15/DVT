```python
import pandas as pd
import seaborn as sns

df = pd.read_csv("medical_dataset_1100_rows_v2.csv")
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["Age"], df["BMI"])
plt.title("Line Plot: Age vs BMI")
plt.xlabel("Age")
plt.ylabel("BMI")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.boxplot(df["BMI"])
plt.title("BMI Box Plot")
plt.ylabel("BMI")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Age"], bins=15, edgecolor="black")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.violinplot(df["BMI"])
plt.title("BMI Violin Plot")
plt.ylabel("BMI")
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df.index, df["Blood_Sugar"], color="red")
plt.title("Blood Sugar Trend")
plt.xlabel("Patient Record")
plt.ylabel("Blood Sugar")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.stackplot(
    df["Patient_ID"],
    df["Height_cm"],
    df["Weight_kg"],
    labels=["Height", "Weight"]
)

plt.title("Height and Weight")
plt.xlabel("Patient ID")
plt.ylabel("Values")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.plot(df["Patient_ID"], df["BMI"])
plt.xticks(rotation=45)
plt.title("BMI by Patient ID")
plt.xlabel("Patient ID")
plt.ylabel("BMI")
plt.tight_layout()
plt.show()
```

```python
disease = df["Disease"].value_counts()

plt.figure(figsize=(7,7))
plt.pie(disease,
        labels=disease.index,
        autopct="%1.1f%%",
        startangle=90)
plt.title("Disease Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Height_cm", y="Weight_kg")
plt.title("Height vs Weight")
plt.show()
```

```python
sns.pairplot(
    df[["Age","Height_cm","Weight_kg","BMI","Blood_Sugar"]]
)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="Age",
    y="Blood_Sugar",
    hue="Gender"
)
plt.title("Age vs Blood Sugar")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(
    data=df,
    x="Gender",
    y="BMI"
)
plt.title("BMI by Gender")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(
    data=df,
    x="Disease"
)
plt.title("Disease Count")
plt.xticks(rotation=45)
plt.show()
```

```python
fig, ax = plt.subplots(1,3, figsize=(15,5))

sns.scatterplot(
    data=df,
    x="Height_cm",
    y="Weight_kg",
    ax=ax[0]
)
ax[0].set_title("Scatter")

sns.boxplot(
    data=df,
    y="BMI",
    ax=ax[1]
)
ax[1].set_title("Box")

sns.countplot(
    data=df,
    x="Gender",
    ax=ax[2]
)
ax[2].set_title("Count")

plt.tight_layout()
plt.show()
```
